```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('선박입출항.csv', encoding='cp949')
df[df['선박종류명'] == '풀컨테이너선']
```

```python
df['출항예정일시'] = pd.to_datetime(df['출항예정일시'], errors='coerce')
df['출항일시'] = pd.to_datetime(df['출항일시'], errors='coerce')

df['출항차이시간'] = (
    df['출항일시'] - df['출항예정일시']
).dt.total_seconds() / 3600
```

```python
df[['업체코드', '차항지명', '적재톤수', '입항일시','출항예정일시','출항일시','양하톤수','출항차이시간','선박종류명']]
```

```python
df = df[df['출항차이시간'] > 0]
df
```

```python
df_container = df[df['선박종류명'] == '풀컨테이너선']
df_container
```

```python
df = df_container[
    (df_container['출항차이시간'] > 1) &
    (df_container['출항차이시간'] < 72)
]
df
```
